```python
import numpy as np
from pyscf import lib

def read_one_electron_integrals(file_name, n_ao):
    integrals = np.zeros((n_ao, n_ao))
    with open(file_name, 'r') as f:
        next(f)  # Skip the header line
        for line in f:
            i, j, val = line.split()
            i, j = int(i)-1, int(j)-1
            integrals[i, j] = float(val)
    return integrals

def read_two_electron_integrals(file_name, n_ao):
    integrals = np.zeros((n_ao, n_ao, n_ao, n_ao))
    with open(file_name, 'r') as f:
        next(f)  # Skip the header line
        for line in f:
            i, j, k, l, val = line.split()
            i, j, k, l = int(i)-1, int(j)-1, int(k)-1, int(l)-1
            integrals[i, j, k, l] = float(val)
    return integrals

def read_overlap_integrals(file_name, n_ao):
    return read_one_electron_integrals(file_name, n_ao)

def transform_integrals(h1, eri, s):
    s_inv = np.linalg.inv(s)
    num_orbitals = h1.shape[0]

    # Transform one-electron integrals
    h1_transformed = np.einsum('pr,rq->pq', s_inv, h1)

    # Transform two-electron integrals
    eri_transformed = np.einsum('pi,qj,ijkl->pqkl', s_inv, s_inv, eri)

    return h1_transformed, eri_transformed

# Set the number of atomic orbitals for H2
n_ao = 2

# Read the integrals from the files
overlap_integrals = read_overlap_integrals('/home/bruna/overlap_integrals_H2.txt', 2 * n_ao)
one_electron_integrals = read_one_electron_integrals('/home/bruna/one_electron_integrals_H2.txt', 2 * n_ao)
two_electron_integrals = read_two_electron_integrals('/home/bruna/two_electron_integrals_H2.txt', 2 * n_ao)

# Transform the integrals
h1_transformed, eri_transformed = transform_integrals(one_electron_integrals, two_electron_integrals, overlap_integrals)

# Write the transformed one-electron integrals to a file
with open("transformed_one_electron_integrals_H2.txt", "w") as f:
    f.write("Transformed One-electron integrals:\n")
    for i in range(2 * n_ao):
        for j in range(2 * n_ao):
            f.write(f"{i+1:5d} {j+1:5d} {h1_transformed[i, j]:20.10E}\n")

# Write the transformed two-electron integrals to a file
with open("transformed_two_electron_integrals_H2.txt", "w") as f:
    f.write("Transformed Two-electron integrals:\n")
    for i in range(2 * n_ao):
        for j in range(2 * n_ao):
            for k in range(2 * n_ao):
                for l in range(2 * n_ao):
                    f.write(f"{i+1:5d} {j+1:5d} {k+1:5d} {l+1:5d} {eri_transformed[i, j, k, l]:20.10E}\n")
```
