```python
import numpy as np
from pyscf import gto, scf

# Define the H2 molecule with STO-3G basis set
mol = gto.M(
    atom='H 0 0 0; H 0 0 0.7414',  # H2 molecule with a bond length of 0.7414 angstroms
    basis='sto-3g'
)

mf = scf.RHF(mol)
mf.kernel()

# Number of atomic orbitals
n_ao = mol.nao

s = mol.intor('int1e_ovlp')
t = mol.intor('int1e_kin')
v = mol.intor('int1e_nuc')
h1 = t + v

# Create the spin orbital matrix for overlap integrals
s_spin = np.zeros((2 * n_ao, 2 * n_ao))

for i in range(n_ao):
    for j in range(n_ao):
        # Alpha-alpha block
        s_spin[i, j] = s[i, j]
        # Beta-beta block
        s_spin[i + n_ao, j + n_ao] = s[i, j]

# Print the overlap integrals
print("Overlap integrals:")
for i in range(2 * n_ao):
    for j in range(2 * n_ao):
        print(f"{i+1:5d} {j+1:5d} {s_spin[i, j]:20.10E}")

# Create the spin orbital matrix for one-electron integrals
h1_spin = np.zeros((2*n_ao, 2*n_ao))

for i in range(n_ao):
    for j in range(n_ao):
        # Alpha-alpha block
        h1_spin[i, j] = h1[i, j]
        # Beta-beta block
        h1_spin[i + n_ao, j + n_ao] = h1[i, j]

# Print the one-electron integrals
print("One-electron integrals:")
for i in range(2 * n_ao):
    for j in range(2 * n_ao):
        print(f"{i+1:5d} {j+1:5d} {h1_spin[i, j]:20.10E}")

# Two-electron integrals in chemist's notation (ik|jl)
eri_chemist = mol.intor('int2e')

# Convert to physicist's notation (ijkl)
def convert_to_physicist(eri_chemist, n_ao):
    eri_physicist = np.zeros((n_ao, n_ao, n_ao, n_ao))
    for i in range(n_ao):
        for j in range(n_ao):
            for k in range(n_ao):
                for l in range(n_ao):
                    eri_physicist[i, j, k, l] = eri_chemist[i, k, j, l]
    return eri_physicist

# Convert to physicist notation
eri_physicist = convert_to_physicist(eri_chemist, n_ao)

# Create the spin orbital matrix for two-electron integrals
eri_spin = np.zeros((2*n_ao, 2*n_ao, 2*n_ao, 2*n_ao))

for i in range(n_ao):
    for j in range(n_ao):
        for k in range(n_ao):
            for l in range(n_ao):
                # Alpha-alpha-alpha-alpha block
                eri_spin[i, j, k, l] = eri_physicist[i, j, k, l]
                # Beta-beta-beta-beta block
                eri_spin[i + n_ao, j + n_ao, k + n_ao, l + n_ao] = eri_physicist[i, j, k, l]
                # Alpha-beta-alpha-beta block
                eri_spin[i, j + n_ao, k, l + n_ao] = eri_physicist[i, j, k, l]
                # Beta-alpha-beta-alpha block
                eri_spin[i + n_ao, j, k + n_ao, l] = eri_physicist[i, j, k, l]

# Print the two-electron integrals
print("\nTwo-electron integrals:")
for i in range(2*n_ao):
    for j in range(2*n_ao):
        for k in range(2*n_ao):
            for l in range(2*n_ao):
                #if abs(eri_spin[i, j, k, l]) > 1e-10:  # Print only significant values
                 print(f"{i+1:5d} {j+1:5d} {k+1:5d} {l+1:5d} {eri_spin[i, j, k, l]:20.10E}")
```
